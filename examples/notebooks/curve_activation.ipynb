# PhotonAct curve activation
This notebook uses the bundled **synthetic** `sample_phh` curve. It is not experimental or paper data.

```python
import torch

from photonact import CurveActivation

layer = CurveActivation.from_file("../../examples/curves/sample_phh.csv", branch="up")
x = torch.linspace(0, 2, 101, requires_grad=True)
y = layer(x)
(gradient,) = torch.autograd.grad(y.sum(), x)
x[:3], y[:3], gradient[:3]
```
